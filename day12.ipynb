```python
import cv2
import matplotlib.pyplot as plt
import numpy as np
```

```python
img=cv2.imread("soccer.bmp",0)
plt.imshow(img,vmin=0,vmax=255,cmap="gray")
plt.show()
```

we use function cv2.threshold()
it takes arguments src image, threshold, what to do with pixels greater than threshold(maxval), type of thresholding 
it returns threshold and output image

```python
ret,bin=cv2.threshold(img,100,255,cv2.THRESH_BINARY)

plt.figure(figsize=(15,15))
plt.subplot(1,2,1)
plt.imshow(img,cmap="gray",vmin=0,vmax=255)
plt.title("Original Image")

plt.subplot(1,2,2)
plt.imshow(bin,cmap="gray",vmin=0,vmax=255)
plt.title("Image after thresholding")

plt.show()
```

Types of thresholding

1 cv2.THRESH_BINARY: greater than a specific value 255(maxval) else 0 
2 cv2.THRESH_BINARY_INV: greater than a specific value 0 else 255(maxval) 
3 cv2.THRESH_TRUNC: greater than threshold is set to threshold 
4 cv2.THRESH_TOZERO: less than threshold is set to 0
5 cv2.THRESH_TOZERO_INV: greater than threshold is set 0

```python
ret,bin=cv2.threshold(img,100,255,cv2.THRESH_BINARY)
ret,inv_bin=cv2.threshold(img,100,255,cv2.THRESH_BINARY_INV)
ret,trunc=cv2.threshold(img,100,255,cv2.THRESH_TRUNC)
ret,setzero=cv2.threshold(img,100,255,cv2.THRESH_TOZERO)
ret,inv_setzero=cv2.threshold(img,100,255,cv2.THRESH_TOZERO_INV)

plt.figure(figsize=(15,15))
plt.subplot(3,2,1)
plt.imshow(img,cmap="gray",vmin=0,vmax=255)
plt.title("Original Image")

plt.subplot(3,2,2)
plt.imshow(bin,cmap="gray",vmin=0,vmax=255)
plt.title("THRESH_BINARY")

plt.subplot(3,2,3)
plt.imshow(inv_bin,cmap="gray",vmin=0,vmax=255)
plt.title("THRESH_BINARY_INV")

plt.subplot(3,2,4)
plt.imshow(trunc,cmap="gray",vmin=0,vmax=255)
plt.title("THRESH_TRUNC")

plt.subplot(3,2,5)
plt.imshow(setzero,cmap="gray",vmin=0,vmax=255)
plt.title("THRESH_TOZERO")

plt.subplot(3,2,6)
plt.imshow(inv_setzero,cmap="gray",vmin=0,vmax=255)
plt.title("THRESH_TOZERO_INV")

plt.show()
```

Problem with threshold is it changes with image to image and it should be dynamic based on statistics of pixel intensity
OTSU takes care of this

```python
ret,otsu_bin=cv2.threshold(img,0,255,cv2.THRESH_BINARY+cv2.THRESH_OTSU)

plt.figure(figsize=(15,15))
plt.subplot(1,3,1)
plt.imshow(img,cmap="gray",vmin=0,vmax=255)
plt.title("Original Image")

plt.subplot(1,3,2)
plt.imshow(bin,cmap="gray",vmin=0,vmax=255)
plt.title("THRESH_BINARY")

plt.subplot(1,3,3)
plt.imshow(otsu_bin,cmap="gray",vmin=0,vmax=255)
plt.title("THRESH_OTSU")

plt.show()
print(ret)
```
