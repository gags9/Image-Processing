Day8 Understanding Derivatives

Derivative is use to determine the change in particular parameter
In Image processing an equivalent is to find the change in pixel intesity 

Example 
[90 90 90] No gradient in image 90-90=0
[20 60 100] increasing gradient 100-20=80
essentially what is being done is  scalar multiplication with [-1 0 1]

Using this concept there are various filters:
Prewitt - gives equal weightage to all pixels 
[
[-1 0 1]
[-1 0 1]
[-1 0 1]
]
Sobel - gives greater weightage to nearby pixels
[
[-1 0 1]
[-2 0 2]
[-1 0 1]
]

The centre is not used as we want to know the change along the centre pixel
Transpose of this is used for gradient in y

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
img=cv2.imread("soccer.bmp",0)
plt.imshow(img,cmap='gray',vmin=0,vmax=255)
plt.show()
```

```python
PrewittX=np.array([[-1,0,1],[-1,0,1],[-1,0,1]])
PrewittY=np.transpose(PrewittX)
Gx=cv2.filter2D(img,cv2.CV_32F,PrewittX)
Gy=cv2.filter2D(img,cv2.CV_32F,PrewittY)
```
