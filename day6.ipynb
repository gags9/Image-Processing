```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
img=cv2.imread("soccer.bmp",0)
plt.imshow(img,cmap='gray',vmin=0,vmax=255)
plt.show()
```

```python
kernelGaussian=cv2.getGaussianKernel(5,1)
print(kernelGaussian)
```

```python
#Gaussian kernel
GuassianBlur=cv2.GaussianBlur(img,(5,5),1)
# 1st argument - source image
# 2nd argument - size of the gaussian kernel(should be odd)
# 3rd argument - standard deviation(if 0 it will automatically take as per kernel size) 

plt.subplot(1,2,1)
plt.imshow(img,cmap='gray',vmin=0,vmax=255)
plt.title('original image')

plt.subplot(1,2,2)
plt.imshow(GuassianBlur,cmap='gray',vmin=0,vmax=255)
plt.title("Gaussian blur")

plt.show()
```

When to use Gaussian filter
- Removing Gaussian noise.
- Preprocessing before edge detection.
- Smoothing images before feature extraction.
- Reducing high-frequency noise

How Gaussian blur varies with different sigma and kernel size 

```python
gb3=cv2.GaussianBlur(img,(3,3),3)
gb5=cv2.GaussianBlur(img,(9,9),3)
gb7=cv2.GaussianBlur(img,(15,15),3)

plt.figure(figsize=(15,15))
plt.subplot(2,2,1)
plt.imshow(img,cmap='gray',vmin=0,vmax=255)
plt.title('Original image')

plt.subplot(2,2,2)
plt.imshow(gb3,cmap='gray',vmin=0,vmax=255)
plt.title('Gaussian Blur with kernel 3')

plt.subplot(2,2,3)
plt.imshow(gb5,cmap='gray',vmin=0,vmax=255)
plt.title('Gaussian Blur with kernel 5')

plt.subplot(2,2,4)
plt.imshow(gb7,cmap='gray',vmin=0,vmax=255)
plt.title('Gaussian Blur with kernel 7')

```

```python
std1=cv2.GaussianBlur(img,(9,9),1)
std3=cv2.GaussianBlur(img,(9,9),3)
std5=cv2.GaussianBlur(img,(9,9),5)

plt.figure(figsize=(15,15))
plt.subplot(2,2,1)
plt.imshow(img,cmap='gray',vmin=0,vmax=255)
plt.title('Original image')

plt.subplot(2,2,2)
plt.imshow(std1,cmap='gray',vmin=0,vmax=255)
plt.title('Gaussian Blur with standard deviation 1')

plt.subplot(2,2,3)
plt.imshow(std3,cmap='gray',vmin=0,vmax=255)
plt.title('Gaussian Blur with standard deviation 3')

plt.subplot(2,2,4)
plt.imshow(std5,cmap='gray',vmin=0,vmax=255)
plt.title('Gaussian Blur with standard devaition 7')

plt.show()
```
