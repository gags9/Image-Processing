Edge detection

```python
import cv2
import matplotlib.pyplot as plt
import numpy as np
```

```python
img=cv2.imread("soccer.bmp",0)

plt.imshow(img,cmap='gray',vmin=0,vmax=255)
plt.tilte("Original image")
plt.show()
```

```python
#Sobel Edge detection

gx=cv2.Sobel(img,cv2.CV_64F,1,0,ksize=3)
gy=cv2.Sobel(img,cv2.CV_64F,0,1,ksize=3)

plt.figure(figsize=(15,15))

plt.subplot(1,2,1)
plt.imshow(gx,cmap="gray",vmin=0,vmax=255)
plt.title("gradient in x")

plt.subplot(1,2,2)
plt.imshow(gx,cmap="gray",vmin=0,vmax=255)
plt.title("gradient in y")


plt.show()
```

```python
#combine x gradient and y gradient
net=cv2.magnitude(gx,gy)
plt.imshow(net,cmap="gray",vmin=0,vmax=255)
plt.title("gradient Magnitude")
plt.show()
```

```python
#Comparing different ksize of sobel

gx3=cv2.Sobel(img,cv2.CV_64F,1,0,ksize=1)
gx7=cv2.Sobel(img,cv2.CV_64F,1,0,ksize=3)
gx11=cv2.Sobel(img,cv2.CV_64F,1,0,ksize=5)

plt.figure(figsize=(15,15))

plt.subplot(1,3,1)
plt.imshow(gx3,cmap="gray",vmin=0,vmax=255)
plt.title("gradient in x with ksize=3")

plt.subplot(1,3,2)
plt.imshow(gx7,cmap="gray",vmin=0,vmax=255)
plt.title("gradient in x with ksize=7")

plt.subplot(1,3,3)
plt.imshow(gx11,cmap="gray",vmin=0,vmax=255)
plt.title("gradient in x with ksize=11")

plt.show()
```
